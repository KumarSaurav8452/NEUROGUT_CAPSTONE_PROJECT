```python
import pandas as pd
import numpy as np

def generate_frontend_test_data(n_samples=50):
    biomarkers = [
        'Lachnospira', 'Ruthenibacterium', 'Veillonella', 'Gardnerella',
        'Pseudobutyrivibrio', 'Kocuria', 'Trueperella', 'Lancefieldella',
        'Fructilactobacillus', 'Lachnoanaerobaculum', 'Corynebacterium',
        'Raoultella', 'Lawsonella'
    ]
    
    # Create base data with random noise (representing natural microbiome variance)
    data = np.random.uniform(0.1, 0.4, size=(n_samples, len(biomarkers)))
    df = pd.DataFrame(data, columns=biomarkers)
    
    y_true = []
    for i in range(n_samples):
        # Assign clinical groups for validation testing
        # Every 4th patient is 'Comorbid'
        if i % 4 == 0:
            df.at[i, 'Raoultella'] += 0.5  # Risk Factor High
            df.at[i, 'Lancefieldella'] += 0.4 # Risk Factor High
            df.at[i, 'Lachnospira'] -= 0.1 # Protective Factor Low
            y_true.append([1, 1])
        # Every 4th + 1 is 'Depression only'
        elif i % 4 == 1:
            df.at[i, 'Raoultella'] += 0.5
            y_true.append([1, 0])
        # Every 4th + 2 is 'Anxiety only'
        elif i % 4 == 2:
            df.at[i, 'Lancefieldella'] += 0.5
            y_true.append([0, 1])
        # Others are 'Healthy'
        else:
            y_true.append([0, 0])
            
    return df, np.array(y_true)

X_synthetic, y_synthetic = generate_frontend_test_data()
X_synthetic.to_csv('frontend_validation_input.csv', index=False)
```
